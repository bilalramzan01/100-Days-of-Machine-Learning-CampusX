## Feature Scaling or Standardization

Feature Scaling is the process of transforming numeric features so they all fall within a similar range or have similar statistical properties — preventing features with larger numeric ranges from unfairly dominating your model (we actually covered the why of this earlier when you asked about scaling for logistic regression).

### Why it's needed (quick recap)
If one feature is "Age" (18–70) and another is "Salary" (20,000–200,000), algorithms using distance or gradient descent will treat "Salary" as far more important samply because its numbers are bigger — not because it's actually more predictive.

### Types of feature scalling
1. Standardization
2. Normalization

#### Standardization (Z-Score Normalization)

Standardization (Z-Score Normalization) transforms your data so it has a mean of 0 and a standard deviation of 1 — essentially reshaping the data around a common center point without changing its underlying distribution shape.

***The formula***
z = (x - μ) / σ

### Where:

- x = original value
- μ (mu) = mean of the column
- σ (sigma) = standard deviation of the column
- z = the standardized value

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df = df.iloc[: , 2:]
```

```python
df.sample(5)
```

```python
df.head(5)
```

### Train test split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop('Purchased', axis=1), df['Purchased'], test_size=0.30, random_state=0)
```

```python
x_train.shape, x_test.shape
```

### Standard Scaller

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
# fit the scalar
scaler.fit(x_train)
```

```python
# Transoform train and test set
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns = x_train.columns)
```

```python
np.round(x_train_scaled.describe(), 1)
```

```python
np.round(x_train.describe(), 1)
```

### why Scalling is impportant ?

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr_scaled = LogisticRegression()

lr.fit(x_train, y_train)
lr_scaled.fit(x_train_scaled, y_train)

y_pred = lr.predict(x_test)
y_pred_scaled = lr_scaled.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

print("Actual", accuracy_score(y_test, y_pred))
```

```python
print("Scaled", accuracy_score(y_test, y_pred_scaled))
```

**Must use standardization of Feature scaling when these algorithms used.**

- K-means
- KNN
- PCA
- ANN
- Gradient descent

**When not to use scalling**
- Decision tree
- Random forest
- Gradient Boost
- XG Boost
