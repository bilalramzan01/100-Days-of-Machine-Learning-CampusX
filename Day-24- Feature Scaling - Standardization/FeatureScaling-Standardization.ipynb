## Feature Scaling or Standardization

Feature Scaling is the process of transforming numeric features so they all fall within a similar range or have similar statistical properties — preventing features with larger numeric ranges from unfairly dominating your model (we actually covered the why of this earlier when you asked about scaling for logistic regression).

### Why it's needed (quick recap)
If one feature is "Age" (18–70) and another is "Salary" (20,000–200,000), algorithms using distance or gradient descent will treat "Salary" as far more important simply because its numbers are bigger — not because it's actually more predictive.

### Types of feature scalling
1. Standardization
2. Normalization
