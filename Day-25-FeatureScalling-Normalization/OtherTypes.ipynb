## Mean Normalization
Mean Normalization is a feature scaling technique that rescales data into a range centered around 0, using the mean instead of the minimum value (unlike Min-Max Normalization, which centers around the minimum).

## Max Absolute Normalization
Max Absolute Normalization (MaxAbs Scaling) rescales each value by dividing it by the maximum absolute value in that column — squeezing all values into the range [-1, 1], while preserving the sign (positive/negative) of the original data.

## Robust Scalling
Robust Scaling is a feature scaling technique specifically designed to handle outliers — instead of using mean/std (Standardization) or min/max (Normalization), which outliers heavily distort, it uses the median and IQR (Interquartile Range), which are much more resistant to extreme values.
