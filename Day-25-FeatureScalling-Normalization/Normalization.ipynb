### Feature Scaling - Normalization | MinMaxScaling | MaxAbsScaling | RobustScaling

Normalization is a feature scaling technique applied during data preparation for machine learning. Its goal is to rescale the values of numeric columns onto a common scale (typically 0 to 1), without distorting the differences between value ranges or losing information about the relative relationships in the data.

### Types of Normalization.

1. MinMax Scalling
2. Mean normalization
3. maxAbs Scalling
4. Robust scalling

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('wine_data.csv', header=None, usecols=[0,1,2])
```

```python
df.columns = ['Class label', 'Alcohol', 'Malic Acid']
```

```python
df.head(5)
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
sns.kdeplot(df['Malic Acid'])
```

```python
sns.scatterplot(x=df['Alcohol'], y=df['Malic Acid'], hue=df['Class label'], palette=['Red','Green','Purple'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop('Class label', axis=1), df['Class label'],test_size=0.3,random_state=0)
```

```python
x_train.shape, x_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns=x_test.columns)
```
