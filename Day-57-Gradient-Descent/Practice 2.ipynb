```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=10, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x, reg.predict(x), color='red')
```

```python
m = 48.53
b = 100

y_pred = ((48.53*x) + b).reshape(10)
```

```python
plt.scatter(x,y)
plt.plot(x, reg.predict(x), color='red')
plt.plot(x,y_pred, color='green', label='b = {}'.format(b))
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum (y - 48.53*x.ravel() - b)
loss_slope
```

```python
lr = 0.10

step_size = lr * loss_slope
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((48.53*x) + b).reshape(10)
```

```python
plt.scatter(x,y)
plt.plot(x, reg.predict(x), color='red')
plt.plot(x,y_pred1, color='purple', label='b = {}'.format(b))
plt.plot(x,y_pred, color='green', label='b = {}'.format(b))
plt.legend()
plt.show
```
