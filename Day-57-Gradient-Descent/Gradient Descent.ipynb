# Gradient Descent

- Gradient descent is a method for unconstrained mathematical optimization. It is a first-order iterative algorithm for minimizing a differentiable multivariate function.
- It is an algorithm to find minima
- Gradient descent is an alogorithm use to minimize the loss function. Loss function is the difference between the actual value and predicted value.

```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples = 4, n_features = 1, n_informative = 1, n_targets = 1, noise = 80, random_state = 13 )
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x,y)
```

```python
# Lets apply OLS

from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red')
```

```python
# Lets find out the Gradient Descent by assuming slope is constant m = 78.35
# lets suppose the starting value for the y intercept is b = 0
y_pred = ((78.35*x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred,color='green', label = 'b = 0')
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_slope = -2 * np.sum (y - m*x.ravel() - b)
loss_slope
```

```python
# Lets take the learning rate 

lr = 0.1

step_size = loss_slope*lr
step_size
```

```python
# Calculating new b

b = b - step_size
b
```

```python
y_pred1 = ((78.35*x)+b).reshape(4)

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred1,color='purple', label = 'b = {}'.format(b))
plt.plot(x,y_pred,color='green', label = 'b = 0')
plt.legend()
plt.show()
```
