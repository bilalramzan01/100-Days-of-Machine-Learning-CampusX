```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=5, n_informative=1, n_targets=1, n_features=1, noise=80, random_state=13)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x), color='red', label = 'OSL')
plt.legend()
plt.show()
```

```python
# Lets suppose we have m constand m = 74.25
# and we have b = 100

m = 74.25
b = 100
```

```python
y_pred = ((74.25*x) + b).reshape(5)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred,color='green', label = 'b = 100')
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum (y - m*x.ravel() - b)
loss_slope
```

```python
lr = .05
step_size = lr*loss_slope
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((74.25*x) + b)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred1,color='purple', label = 'b = 100')
plt.plot(x,y_pred,color='green', label = 'b = 100')
plt.legend()
plt.show()
```
