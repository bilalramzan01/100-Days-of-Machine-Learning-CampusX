# Scikit-Learn Pipelines

Scikit-Learn Pipelines chain together multiple preprocessing steps (and a final model) so that the output of each step automatically becomes the input to the next step — all bundled into a single object.

**Why this matters (directly connects to your last error!)**

You just hit a real bug caused by manually applying SimpleImputer separately to xtrain and xtest — remember the data leakage issue we just fixed (fit_transform on train, transform on test)? Pipelines solve this exact problem automatically.
